<FONT COLOR ="MAGENTA"><H2><B><U><I>Remove Outliers using IQR Method</I></U></B></H2></FONT>

```python
# Function to remove outliers
def remove_outliers(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df[col] = np.where(df[col] < lower, lower, df[col])
        df[col] = np.where(df[col] > upper, upper, df[col])
    return df

df_clean = remove_outliers(df.copy(), num_cols)
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Encode Categorical Features</I></U></B></H2></FONT>

* **Categorical features**: Protocol, Flag, Family, SeedAddress, ExpAddress, IPaddress, Threats, Prediction
* Use LabelEncoder for simplicity (for tree-based models)

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy to encode
df_model = df.copy()

# Initialize label encoder
le = LabelEncoder()

# List of categorical columns to encode
cat_cols = ['Protocol', 'Flag', 'Family', 'SeedAddress', 'ExpAddress', 'IPaddress', 'Threats', 'Prediction']

# Encode each categorical column
for col in cat_cols:
    df_model[col] = le.fit_transform(df_model[col])

# Display first 5 rows after encoding
df_model.head()
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Scale Numeric Features</I></U></B></H2></FONT>

* **Numeric features**: Time, Clusters, BTC, USD, Netflow_Bytes, Port
* Use StandardScaler for normalization.

```python
from sklearn.preprocessing import StandardScaler

# Numeric columns
num_cols = ['Time', 'Clusters', 'BTC', 'USD', 'Netflow_Bytes', 'Port']

# Initialize scaler
scaler = StandardScaler()

# Fit and transform numeric columns
df_model[num_cols] = scaler.fit_transform(df_model[num_cols])

# Display first 5 rows after scaling
df_model.head()
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Explanation:</I></U></B></H2></FONT>

**Encoding**: Converts categorical labels into numeric format for ML models.

**Scaling**: Standardizes numeric values for consistent model performance, especially for distance-based algorithms (like KNN or SVM).

Now, df_model is ready for classification model building.

<FONT COLOR ="BLUE"><H1><B><U><I>Model Building</I></U></B></H1></FONT>


<FONT COLOR ="MAGENTA"><H2><B><U><I>Split Dataset into Features and Target</I></U></B></H2></FONT>

```python
from sklearn.model_selection import train_test_split

# Features (X) and target (y)
X = df_model.drop('Prediction', axis=1)
y = df_model['Prediction']

# Split into training and testing sets (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Display shapes
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Import Classifiers</I></U></B></H2></FONT>

 We will test multiple classifiers to compare performance:

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# For evaluation
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Initialize Models</I></U></B></H2></FONT>

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=500),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "KNN": KNeighborsClassifier(),
    "SVM": SVC()
}
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Train, Predict & Evaluate Models</I></U></B></H2></FONT>

```python
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    results[name] = acc

    print(f"--- {name} ---")
    print("Accuracy:", acc)
    print("Classification Report:\n", classification_report(y_test, y_pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
    print("\n")
```

<FONT COLOR ="MAGENTA"><H2><B><U><I>Compare Model Accuracy</I></U></B></H2></FONT>

```python
# Display comparison
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.bar(results.keys(), results.values(), color='teal')
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")
plt.xticks(rotation=45)
plt.show()
```

- KNN is good but slower and slightly less accurate than trees.
- SVM performs reasonably but struggles with the largest class (Safe) and is computationally expensive.

<FONT COLOR="BLUE"><H1><B><U><I>Hyperparameter Tuning & Final Model</I></U></B></H1></FONT>

```python
# Import Required Libraries
import time
from sklearn.model_selection import RandomizedSearchCV
```

<FONT COLOR="BLUE"><H1><B><U><I>Define Hyperparameter Grid</I></U></B></H1></FONT>

<FONT COLOR="MAGENTA"><H2><B><U><I>Configure and run RandomizedSearchCV (fast)</I></U></B></H1></FONT> </FONT>
* n_iter :  controls how many random combinations are tested (lower = faster).
* cv=2 :  uses 2-fold CV for speed.
* Start the search and time it so you can report runtime.

```python

# Initialize the Random Forest classifier
rf = RandomForestClassifier(random_state=42)

# Define the hyperparameter grid for RandomizedSearchCV
param_dist = {
    'n_estimators': [50, 100, 150, 200],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}
# Define base model
rf = RandomForestClassifier(random_state=42)
random_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=15,            # try 15 random combinations (fast)
    cv=2,                 # 2-fold CV for speed
    n_jobs=-1,
    verbose=2,
    scoring='accuracy',
    random_state=42
)

start = time.time()
random_search.fit(X_train, y_train)
end = time.time()
print(f"RandomizedSearchCV done in {(end-start)/60:.2f} minutes")
print("Best Parameters:", random_search.best_params_)
print("Best CV Accuracy:", random_search.best_score_)
```

<FONT COLOR="MAGENTA"><H2><B><U><I>Train Final Model with Best Hyperparameters</I></U></B></H1></FONT> </FONT>

```python
best_rf = random_search.best_estimator_
best_rf.fit(X_train, y_train)
```

<FONT COLOR="MAGENTA"><H2><B><U><I>Evaluate final model on test set</I></U></B></H1></FONT> </FONT>

```python
y_pred = best_rf.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Confusion matrix plot
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - Final Random Forest')
plt.show()
```
